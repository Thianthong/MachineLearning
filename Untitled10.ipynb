```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.preprocessing import StandardScaler
```

```python
def prep(A):
  return scaler.transform(poly.transform(A))
```

```python
from sklearn.datasets import make_moons
x_train,y_train = make_moons(n_samples = 50 ,noise =0.35, random_state=51)
x_test,y_test = make_moons(n_samples = 500 ,noise =0.35, random_state=105)
print(x_train.shape,x_test.shape)
```

```python
for k in [0,1]:
  pts = x_train[y_train == k]
  plt.scatter(pts[:,0],pts[:,1],label = f"y = {k}")
plt.xlabel("x1")
plt.ylabel("x2")
plt.legend()
plt.show()
```

```python
poly = PolynomialFeatures(degree=10,include_bias=False)
scaler = StandardScaler()
z_train = scaler.fit_transform(poly.fit_transform(x_train))
z_test = prep(x_test)
print(z_train.shape,z_test.shape)
```

```python
def plot_boundary(model, prep, X, y, title=""):
  x1 = np.linspace(X[:, 0].min() - 0.5, X[:, 0].max() + 0.5, 200)
  x2 = np.linspace(X[:, 1].min() - 0.5, X[:, 1].max() + 0.5, 200)
  xx, yy = np.meshgrid(x1, x2)
  Z = prep(np.c_[xx.ravel(), yy.ravel()])
  z = model.decision_function(Z)
  for k in [0, 1]:
    pts = X[y == k]
    plt.scatter(pts[:, 0], pts[:, 1], label=f"y = {k}")
  plt.contour(xx, yy, z.reshape(xx.shape), levels=[0], colors="k")
  plt.xlabel("x1")
  plt.ylabel("x2")
  plt.title(title)
  plt.legend()
  plt.show()
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def cost_function(X, y, w, b, lam):
    m = len(y)

    h = sigmoid(X @ w + b)

    cost = -np.mean(
        y*np.log(h) + (1-y)*np.log(1-h)
    )

    reg = (lam/(2*m)) * np.sum(w**2)

    return cost + reg
```

```python
def gradient_descent(X, y, lam, lr=0.05, epochs=10000):
    m, n = X.shape

    w = np.zeros(n)
    b = 0

    for i in range(epochs):
        h = sigmoid(X @ w + b)

        dw = (X.T @ (h-y))/m + (lam/m)*w
        db = np.mean(h-y)

        w -= lr*dw
        b -= lr*db

    return w, b
```

```python
class MyLogisticRegression:

    def __init__(self, penalty=None, lam=0.0, max_iter=10000):
        self.penalty = penalty
        self.lam = lam
        self.max_iter = max_iter

    def fit(self, X, y):
        lam = self.lam if self.penalty == "l2" else 0

        self.coef_, self.intercept_ = gradient_descent(
            X, y,
            lam=lam,
            epochs=self.max_iter
        )

    def decision_function(self, X):
        return X @ self.coef_ + self.intercept_

    def predict(self, X):
        return (self.decision_function(X) >= 0).astype(int)

    def score(self, X, y):
        return np.mean(self.predict(X) == y)
```

```python
model = MyLogisticRegression(penalty=None,max_iter=10000)
model.fit(z_train,y_train)
print("train",model.score(z_train,y_train))
print("test",model.score(z_test,y_test))
print("max weight",round(np.abs(model.coef_).max(),2))
plot_boundary(model,prep,x_train,y_train,"no regu")
```

```python
lams = np.logspace(-3,3,30)
acc_train, acc_test = [],[]
for lam in lams:
  model = MyLogisticRegression(penalty="l2",lam=lam,max_iter=10000)
  model.fit(z_train,y_train)
  acc_train.append(model.score(z_train,y_train))
  acc_test.append(model.score(z_test,y_test))
i = np.argmax(acc_test)
print("best lamda", round(lams[i],3))
print("test acc",acc_test[i])
plt.semilogx(lams,acc_train,label="train")
plt.semilogx(lams,acc_test,label="test")
plt.xlabel("lamda")
plt.xlabel("acc")
plt.legend()
plt.show()
```

```python
model = MyLogisticRegression(penalty="l2",lam=0.001,max_iter=10000)
model.fit(z_train,y_train)
print("train",model.score(z_train,y_train))
print("test",model.score(z_test,y_test))
print("max weight",round(np.abs(model.coef_).max(),2))
plot_boundary(model,prep,x_train,y_train,"regu best lambda")
```
